```python
def predict_business(
    population,
    competition,
    monthly_rent,
    customers,
    budget,
    area,
    business_type
):
    new_data = pd.DataFrame({
        "Population": [population],
        "Competition": [competition],
        "Monthly_Rent": [monthly_rent],
        "Customers": [customers],
        "Budget_Required": [budget],

        "Area_DHA": [area == "DHA"],
        "Area_Hayatabad": [area == "Hayatabad"],
        "Area_Ring Road": [area == "Ring Road"],
        "Area_Saddar": [area == "Saddar"],
        "Area_University Town": [area == "University Town"],

        "Business_Type_Cafe": [business_type == "Cafe"],
        "Business_Type_Gym": [business_type == "Gym"],
        "Business_Type_Pharmacy": [business_type == "Pharmacy"]
    })

    prediction = model.predict(new_data)[0]

    if prediction == 1:
        return "✅ Suitable Business"
    else:
        return "❌ Not Suitable Business"
```

```python
def business_advisor(user_area, user_budget):

    # 1. Filter businesses by area and budget
    options = business_df[
        (business_df["Area"] == user_area) &
        (business_df["Budget_Required"] <= user_budget)
    ].copy()

    if options.empty:
        print("❌ No suitable business found within your budget.")
        return

    # 2. One-hot encoding
    options_encoded = pd.get_dummies(
        options,
        columns=["Area", "Business_Type"]
    )

    # 3. Match model features
    options_encoded = options_encoded.reindex(
        columns=X_encoded.columns,
        fill_value=False
    )

    # 4. ML probability
    options["Suitable_Probability"] = (
        model.predict_proba(options_encoded)[:, 1] * 100
    )

    # 5. Normalize Opportunity Score
    options["Opportunity_Normalized"] = (
        options["Opportunity_Score"]
        / business_df["Opportunity_Score"].max()
        * 100
    )

    # 6. Budget compatibility
    options["Budget_Compatibility"] = (
        (user_budget - options["Budget_Required"])
        / user_budget
        * 100
    )

    # 7. Final ranking score
    options["Final_Ranking_Score"] = (
        options["Suitable_Probability"] * 0.50
        + options["Opportunity_Normalized"] * 0.30
        + options["Budget_Compatibility"] * 0.20
    )

    # 8. Risk score
    options["Risk_Score"] = (
        options["Competition"]
        / business_df["Competition"].max()
        * 100
    )

    # 9. Risk level
    def get_risk(score):
        if score <= 33:
            return "🟢 Low"
        elif score <= 66:
            return "🟡 Medium"
        else:
            return "🔴 High"

    options["Risk_Level"] = options["Risk_Score"].apply(get_risk)

    # 10. Ranking
    ranking = options.sort_values(
        by="Final_Ranking_Score",
        ascending=False
    )

    # 11. Best business
    best = ranking.iloc[0]

    print("🤖 AI BUSINESS ADVISOR")
    print("=" * 45)
    print("📍 Area:", user_area)
    print("💰 Your Budget: Rs.", user_budget)

    print("\n🏆 BEST RECOMMENDATION")
    print("-" * 45)
    print("Business:", best["Business_Type"])
    print("Required Budget: Rs.", best["Budget_Required"])
    print("ML Probability:", round(best["Suitable_Probability"], 2), "%")
    print("Opportunity Score:", round(best["Opportunity_Score"], 2))
    print("Budget Compatibility:", round(best["Budget_Compatibility"], 2), "%")
    print("Final Ranking Score:", round(best["Final_Ranking_Score"], 2))
    print("Competition:", best["Competition"])
    print("Risk Level:", best["Risk_Level"])

    print("\n💡 WHY THIS BUSINESS?")
    if best["Budget_Required"] <= user_budget * 0.70:
        print("✅ Required budget is comfortably within your budget.")

    if best["Suitable_Probability"] >= 70:
        print("✅ High model suitability score.")

    if best["Opportunity_Score"] >= 70:
        print("✅ Strong opportunity score.")

    if best["Competition"] <= 25:
        print("✅ Competition is relatively low.")
    elif best["Competition"] <= 50:
        print("⚠️ Competition is moderate.")
    else:
        print("⚠️ Competition is high.")

    # 12. Top 3
    print("\n🥇 TOP 3 BUSINESSES")
    print("-" * 45)

    top3 = ranking.head(3)

    for rank, (_, row) in enumerate(top3.iterrows(), start=1):
        print(
            f"{rank}. {row['Business_Type']} | "
            f"Score: {row['Final_Ranking_Score']:.2f} | "
            f"ML: {row['Suitable_Probability']:.0f}% | "
            f"Risk: {row['Risk_Level']}"
        )
```
